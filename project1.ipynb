```python
# prompt: Using dataframe df: averge of monthly failure

import pandas as pd
# Convert the timestamp column to datetime objects
df['timestamp'] = pd.to_datetime(df['timestamp'])
# Extract the month and year for grouping
df['month'] = df['timestamp'].dt.to_period('M')
# Group by month and calculate the mean of failure_risk
monthly_failure_average = df.groupby('month')['failure_risk'].mean()
# Print the result
monthly_failure_average
```

```python
# prompt: consumption of motor current

# Filter for rows where motor_current is not NaN
df_motor_current = df.dropna(subset=['motor_current'])
# Group by month and calculate the mean of motor_current
monthly_motor_current_average = df_motor_current.groupby('month')['motor_current'].mean()
# Print the result
monthly_motor_current_average
```
