```python
knn=KNeighborsRegressor().fit(X_train,Y_train)
ada=AdaBoostRegressor().fit(X_train,Y_train)
svm=SVR().fit(X_train,Y_train)
ridge=Ridge().fit(X_train,Y_train)
lasso=Lasso().fit(X_train,Y_train)
rf=RandomForestRegressor().fit(X_train,Y_train)
gbm=GradientBoostingRegressor().fit(X_train,Y_train)
```

```python
models=[ridge,lasso,knn,ada,svm,rf,gbm]
```

```python
def ML(Y,models):
    y_pred=models.predict(X_test)
    mse=mean_squared_error(Y_test,y_pred)
    rmse=np.sqrt(mean_squared_error(Y_test,y_pred))
    r2=r2_score(Y_test,y_pred)*100

    return mse,rmse,r2
```

```python
for i in models:
    print("\n",i,"\n\nDifferent models success rate :",ML("salary_in_usd",i))
```

# Logistic Regression


```python

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Assuming you want to predict 'experience_level' (a categorical variable)
# For logistic regression, the target variable must be categorical.
# Let's use 'experience_level' as the target for demonstration.
# First, we need to ensure it's properly encoded if it's not already.
# In the provided code, 'experience_level' is already factorized, which works.

Y_classification = df['experience_level']
X_classification = df.drop(['experience_level', 'salary_in_usd'], axis=1) # Dropping 'salary_in_usd' as it's not relevant for this classification

X_train_clf, X_test_clf, Y_train_clf, Y_test_clf = train_test_split(X_classification, Y_classification, test_size=0.2, random_state=42, stratify=Y_classification) # Use stratify for classification

print("\nLogistic Regression for Experience Level Prediction:\n")

# Create and train the Logistic Regression model
log_reg = LogisticRegression(max_iter=1000) # Increase max_iter for convergence
log_reg.fit(X_train_clf, Y_train_clf)

# Make predictions on the test set
Y_pred_log_reg = log_reg.predict(X_test_clf)

# Evaluate the model
accuracy = accuracy_score(Y_test_clf, Y_pred_log_reg)
print("Accuracy:", accuracy * 100)

# You can also print a classification report for more detailed metrics
print("\nClassification Report:\n", classification_report(Y_test_clf, Y_pred_log_reg))

# And a confusion matrix
print("\nConfusion Matrix:\n", confusion_matrix(Y_test_clf, Y_pred_log_reg))
```
